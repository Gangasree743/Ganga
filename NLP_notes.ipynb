# 5. RNN

**RNN = Recurrent Neural Network**

RNN processes text **sequentially**, one word/token at a time.

Example:

```text
I → love → this → movie
```

It tries to remember information from previous words.

### Problem

RNN has difficulty remembering information from very long sequences.

This led to models such as **LSTM**.

---

# 6. LSTM

**LSTM = Long Short-Term Memory**

LSTM is an improved type of RNN.

It can remember important information for a longer period.

It uses special gates to control:

* What information to keep
* What information to forget
* What information to output

### Advantage

Better than basic RNN for long-term dependencies.

---

# 7. Transformer

Transformers are a major architecture used in modern NLP.

They use **attention mechanisms** to understand relationships between words.

Example:

> "The animal didn't cross the road because **it** was tired."

The model can use surrounding words to understand what **"it"** refers to.

### Examples of Transformer-based models

* BERT
* GPT
* T5

Transformers are the foundation of many modern NLP systems.

# 11. HTML Tag Removal

Online reviews and web pages can contain HTML tags.

Example:

```html
<p>This movie is good</p>
```

After removing HTML:

```text
This movie is good
```

We commonly use **Regular Expressions (`re`)** for this.

```python
import re

def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

Apply to the dataset:

```python
df['review'] = df['review'].apply(remove_html_tags)
```

# 13. Punctuation Removal

Punctuation includes:

```text
! @ # $ % & * ? . , ; :
```

Example:

```text
"This movie is great!"
```

becomes:

```text
"This movie is great"
```

Python provides:

```python
import string

string.punctuation
```

We can remove punctuation using `translate()`:

```python
exclude = string.punctuation

def remove_punc(text):
    return text.translate(str.maketrans('', '', exclude))
```

# 16. Stopwords

**Stopwords** are common words that often provide limited information for certain NLP tasks.

Examples:

```text
the
is
a
an
of
to
in
and
```

Example:

```text
"This is a good movie"
```

After stopword removal:

```text
"good movie"
```

NLTK provides stopword lists.

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

stopwords.words('english')
```

### Important

Don't blindly remove stopwords in every NLP project.

For example, in **sentiment analysis**, words like **"not"** can be important:

```text
"This movie is not good"
```

Removing `"not"` could change the meaning.

# 19. `split()` Function

The simplest way to perform word splitting in Python is:

```python
sent = "I am going to Delhi"
sent.split()
```

Output:

```text
['I', 'am', 'going', 'to', 'Delhi']
```

### Sentence splitting

```python
sent.split('.')
```

can split sentences based on `.`.

### Problem with `split()`

It does not properly understand language.

For example:

```text
"I have a Ph.D in A.I."
```

There are periods inside abbreviations.

Therefore, specialized tokenizers are generally better.

---

# 20. Regular Expression (`re`) Tokenization

Python's `re` module can find patterns in text.

Example:

```python
import re

sent = "I am going to Delhi!"

tokens = re.findall(r"[\w']+", sent)
```

Output:

```text
['I', 'am', 'going', 'to', 'Delhi']
```

Regex is useful when you need **custom text-processing rules**.
